```python
import pandas as pd
M={"marks" : [45, 78, 65, 65, 88, 54]}
df=pd.DataFrame(M)
df
```

99. Find maximum without using max().

```python
maximum = df["marks"].sort_values(ascending=False).iloc[0]

print("Maximum Marks:", maximum)
```


100. Find minimum without using min().

```python
minimum=df["marks"].sort_values(ascending=True).iloc[0]
print("Minimum Value:", minimum)
```


101. Count values greater than 70 without using count().

```python
Count=df.loc[df["marks"]>70]
print(Count)
```

102. Find the second largest number without using sort().


```python
second_Largest=df.nlargest(2,"marks")["marks"].iloc[1]
print("second_Largest:",second_Largest)
```


103. Find the second smallest number.

```python
second_smallest=df["marks"].sort_values(ascending=True).iloc[1]
print("second_smallest:",second_smallest)
```


104. Find duplicate values.

```python
duplicate=df[df["marks"].duplicated()]
print("duplicate value:",duplicate)           
```

105. Count frequency of each value

```python
df["marks"].value_counts()
```

106. Find the first repeated value.

```python
first=df["marks"].duplicated()
print(df[first].iloc[0])
```

107. Find the first non-repeated value

```python
first=df["marks"].unique()
first[0]
```

```python
even=df["marks"]%2==0
print(df[even])

```

```python
odd=df["marks"]%2==1
print(df[odd])
```
