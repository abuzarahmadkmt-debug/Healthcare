# Group BY:- it is used to join two or more column and give me result.

```python
import pandas as pd

data = {
    "Patient_ID": [101,102,103,104,105,106,107,108,109,110],
    "Age": [25,45,35,60,50,28,65,40,55,30],
    "Gender": ["Male","Female","Male","Female","Male",
               "Female","Male","Female","Male","Female"],
    "Department": ["Cardiology","Neurology","Cardiology","Orthopedic",
                   "Neurology","Cardiology","Orthopedic","Neurology",
                   "Cardiology","Orthopedic"],
    "City": ["Nagpur","Pune","Nagpur","Mumbai","Pune",
             "Nagpur","Mumbai","Pune","Nagpur","Mumbai"],
    "Treatment_Cost": [50000,35000,60000,45000,30000,
                       55000,70000,40000,65000,50000],
    "Hospital_Stay": [5,3,6,4,2,5,7,3,6,4]
}

df = pd.DataFrame(data)

df
```

1. हर city में कितने patients हैं?

```python
df.groupby("City")["Patient_ID"].count()
```

2. हर city की average age क्या है?

```python
df.groupby("City")["Age"].mean()
```

1. Average treatment cost for each department.

```python
df.groupby("Department")["Treatment_Cost"].mean()
```

Q2. Total treatment cost for each department.

```python
df.groupby("Department")["Treatment_Cost"].sum()
```

Q3. Number of patients in each department.

```python
df.groupby("Department")["Patient_ID"].count()
```

Q4. Average hospital stay for each department.

```python
df.groupby("Department")["Hospital_Stay"].mean()
```

Q5. Total treatment cost for each city.

```python
df.groupby("City")["Treatment_Cost"].sum()
```

Q6. Compare average treatment cost of Male and Female patients.

```python
df.groupby("Gender")["Treatment_Cost"].mean()
```

Q7. Number of patients from each city.

```python
df.groupby("City")["Patient_ID"].count()
```

Q8. Average treatment cost for each City + Department combination.

```python
Q13. The hospital manager wants one report containing, for every department: average treatment cost,
total treatment cost, minimum treatment cost and maximum treatment cost. First identify the Group By
column, Analysis column and functions, then write the code.
```

```python
df.groupby(["Department"])["Treatment_Cost"].agg(["mean","sum","max","min"])
```

Q14. Find the average Hospital_Stay for each department and arrange the result from highest to lowest.
Decide which department has the highest average stay.

```python
df.groupby("Department")["Hospital_Stay"].mean().sort_values(ascending=False)
```
