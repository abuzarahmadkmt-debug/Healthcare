```python
import pandas as pd

data = {
    "Patient_ID": ["P101", "P102", "P103", "P104", "P105",
                   "P106", "P107", "P108", "P109", "P110"],

    "Age": [45, 62, 34, 55, 41, 68, 29, 52, 47, 73],

    "City": ["Hyderabad", "Warangal", "Hyderabad", "Vijayawada",
             "Hyderabad", "Nizamabad", "Hyderabad", "Warangal",
             "Hyderabad", "Karimnagar"],

    "Surgery_Type": [
        "Cholecystectomy", "CABG", "Appendectomy",
        "Hernia Repair", "Knee Replacement", "PTCA",
        "Appendectomy", "Cholecystectomy",
        "Hernia Repair", "Knee Replacement"
    ],

    "Operation_Duration_Min": [90, 240, 75, 110, 180,
                               150, 70, 95, 120, 200],

    "Length_of_Stay_Days": [2, 7, 1, 3, 5,
                            4, 1, 2, 3, 6],

    "Admission_Type": [
        "Planned", "Emergency", "Emergency", "Planned",
        "Planned", "Emergency", "Emergency", "Planned",
        "Planned", "Planned"
    ]
}

df = pd.DataFrame(data)

df
```

89. The hospital manager wants to identify high-attention patients.

Conditions:

Age > 50

AND

Operation_Duration_Min > 100

AND

Length_of_Stay_Days > 3

Find and display:

Patient_ID
Age
Surgery_Type
Operation_Duration_Min
Length_of_Stay_Days

```python
df.loc[(df["Age"]>50)&(df["Operation_Duration_Min"]>100)&(df["Length_of_Stay_Days"]>3)][["Patient_ID","Age","Surgery_Type","Operation_Duration_Min","Length_of_Stay_Days"]]
```

90. The hospital wants to review Emergency patients who had long operations.

Conditions:

Admission_Type = Emergency

AND

Operation_Duration_Min > 100

Display:

Patient_ID
Surgery_Type
Operation_Duration_Min

```python
df.loc[(df["Admission_Type"]=="Emergency")&(df["Operation_Duration_Min"]>100)][["Patient_ID","Surgery_Type","Operation_Duration_Min"]]
```

91. Find Hyderabad patients whose operation duration is greater than the average operation duration of the complete dataset.

```python
df.loc[(df["City"]=="Hyderabad")&(df["Operation_Duration_Min"]>df["Operation_Duration_Min"].mean())][["Patient_ID"]]
```
