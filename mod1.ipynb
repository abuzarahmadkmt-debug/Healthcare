```python
df.tail(1)
```


25. Find the length.

```python
df.__len__()
```

26. Find the maximum age.

```python
df.max()
```

27. Find the minimum age.


```python
df.min()
```


28. Find ages greater than 50.


```python
df[df["Age"]>50],"Age"
```

29. Find ages less than 40.

```python
df[df["Age"]<40],"Age"
```

30. Count ages greater than 50.

```python
df[df["Age"]>50].count()
```

31. Calculate average age.

```python
df.mean()
```


32. Check whether 55 exists.

```python
df[df["Age"]==55]
```

33. Find the position of 55.

```python
df[df['Age'] == 55].index
```

34. Find the second largest age.

```python
df["Age"].nlargest(2).iloc[-1]
```

35. Find the second smallest age.

```python
df["Age"].nsmallest(2).iloc[-1]
```
